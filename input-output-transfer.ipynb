# Photon-Number-Resolving Detector Input-Output Transfer

```python
import math
import numpy as np
```

```python
def transfer_input_to_output(
        quantum_efficiency: float = 0.65,
        input_max: int = 10,
):
    """ 
    Give the transfer matrix from input photon number basis e_i to output photon number basis e_o: e_o = M*e_i.
    Follows explicitly linear algebra matrix and column vector form.
    """
    eta = quantum_efficiency
    if eta < 0 or eta > 1:
        raise ValueError(f"Quantum efficiency must be between 0 and 1. For now it is: {eta}")
    
    n_max = int(input_max)

    transfer_matrix = np.array([
        [math.comb(n,m) * eta**m * (1-eta)**(n-m)
         for n in range(n_max + 1)]
         for m in range(n_max + 1)
    ])
    return transfer_matrix

M = transfer_input_to_output(quantum_efficiency=0.1, input_max=5)
v_i = np.array([[0],[1],[0],[0],[0],[0]])
print(M)
M @ v_i
Minv = np.linalg.inv(M)
Minv @ np.array([[.729],[0.243],[.027],[.001],[0],[0]])
```
